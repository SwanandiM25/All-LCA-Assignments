```python
#ASSIGNMENT 2 - program to find largest of 3 numbers
a = int(input("Enter a = "))
b = int(input("Enter b = "))
c = int(input("Enter c = "))

if a > b and a > c:
    print("Largest number is =", a)
elif b > a and b > c:
    print("Largest number is =", b)
else:
    print("Largest number is =", c)
```
